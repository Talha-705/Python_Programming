## List

```python
#Defining and printing
fruits = ['apple', 'banana', 'cherry', 'mango', 'orange', 'almod', 'melon']
print(fruits)
```

# Indexing

```python
fruits[0] 
```

```python
fruits[0:5]
```

```python
fruits[1:5:2]
```

# Negative Indexing

```python
fruits[-1]
```

```python
fruits[-5:-1]
```

# Slicing
[start(default = 0) : end : step (default = 1)]

```python
# Case no: 1
fruits[2: 5]
```

```python
# Case no: 2
fruits[:4] 
```

```python
# Case no: 3
fruits[2:]
```

```python
# Case no: 4
fruits[0:7:2]
```

```python
# Case no: 5
fruits[-5: -2]
```

# 4. Updating list


```python
# case No 1
fruits[2] = 'Watermelon'
print(fruits)
```

```python
#Second Update
fruits[2: 5] = ['cherry', 'stawberry', 'Guava']
print(fruits)
```

# 5. List Manipulation Functions

```python
# insert() 
fruits.insert(2, 'grapes')
print(fruits)
```

```python
# appened() 
fruits.append('lemon')
print(fruits)
```

```python
# extend()

product1 = ['Mobile', 'telephone', 'airbods']
product2 = ['shirts', 'hoddies', 'jeans']

product1.extend(product2)
print(product1)
```

```python
# remove()
fruits.remove('melon')
print(fruits)
```

```python
# pop()
x = fruits.pop(2)
print(fruits)
print("Removed item: ", x)
```

# 6. Sorting

```python
fruits = ['guava', 'Strawbery', 'almod', 'apple', 'grapes']
```

```python
# Case no: 1- sorting 
fruits.sort()
print(fruits)
```

```python
# Case no: 2- sorting in reverse order
fruits.sort(reverse=True)
print(fruits)
```

# 7. List Aliasing

```python
original = fruits
fruits.append('melon')
```

```python
print(fruits)
print(original)
```

```python
original2 = fruits.copy()
fruits.append('cherry')
print(fruits)
print(original2)
```

# 8. Advance Lists (2D Lists, 3D Lists)
## 1. 2D List

```python
# 2D List
fruits_veg = [['apple', 'banana', 'cherry'], ['Brinjal', 'Potato', 'Radish']]
print(fruits_veg)
```

```python
# indexing
print(fruits_veg[1][1])
print(fruits_veg[0][2])
```

```python
# Indexing Multiple Elements
print(fruits_veg[1][1],fruits_veg[0][0])
```

```python
# Slicing a Sublist
print(fruits_veg[1][1:3])
```

```python
# Negative Indexing
print(fruits_veg[-1][-1])
```

# 2. 3D List

```python
fruits_veg = [[['apple', 'banana', 'cherry']],
               [['Brinjal', 'Potato', 'Radish']]]
```

```python
print(fruits_veg[0])
print(fruits_veg[1][0][2])
```
