# DEcorators
A decorator in Python is a function that wraps another function to add extra behavior to it without changing the original function's code.

You attach it using the @decorator_name syntax right above a function definition.

```python
# This is the decorator function
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
# This is the function we want to decorate
@my_decorator
def say_hello():
    print("Hello!")
```

```python
# Call the decorated function
say_hello()
```
