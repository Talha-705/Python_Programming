## String Methods

```python
# length of string
print('Lenght of string:', len(name))
```

```python
# uper case
print('string in Upper case:', name.upper())
```

```python
# lower case
print('string in lower case:', name.lower())
```

```python
# spliting case
word=name.split()
word
```

```python
# joinoig words
print('joinig of words :', '-'.join(word))
```

```python
# finding the index of character
print('Finding the index of character:', name.find('Talha'))
```

```python
# Replace
print('Replacing the character:', name.replace('Talha', 'Ali'))
```

```python
# checking text starts with
print('Checking text starts with:', name.startswith('Muhammad'))
```

```python
# checking text ends with
print('Checking text ends with:', name.endswith('Talha'))
```

```python
# Reverse a string
print('Reverse a string:', name[::-1])
```

```python
# string to list
list1 = list(name)
list1
```

```python
# checkind subsstring presence
print('Checking substring presence:', 'Talha' in name)
```

```python
# string formatting style 1
age=22
print('My name is %s and my age is %d'% (name, age))
```

```python
# srting formating style 2
print('My name is {} and my age is {}'.format(name, age))
```

```python
# string Comparision
str1 = "apple"
str2 = "banana"
is_equal = str1 == str2  
print("Checking strings are equal or not:", is_equal)
is_less = str1 < str2
print("Checking string is less than other string:", is_less)  
```
