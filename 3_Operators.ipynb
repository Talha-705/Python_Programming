```python
a = 10
b = 23
print(a + b) # addition
```

```python
# Subtraction
print(a - b)
```

```python
 # Multiplication
print(a * b) 
```

```python
# Division
print(a / b) 
```

```python
# Remainder Operator(Modulus)
print(a % b) 
```

```python
# Integer Division(Gives answer in Integer)
print(a // b) 
```

```python

# Comparison Operators
# == (Equal)
print(a == b) # (False)

# != (Not Equal)
print(a != b) # (True)

# > (Greater than)
print(a > b) # (False)

# < (Less than)
print(a < b) # (True)

# >= (Greater than or equal to)
print(a >= b) # (False)

# <= (Less than or equal to)
print(a <= b) # (True) 

```

```python

# Logical Operators
# and
# or
# not

# And Operator
print(a > 5 and b > 20) # (True) if both are true.
print(a > 5 and b < 20) # (False) if anyone is false.

```

```python
# OR Operator
print(a > 5 or b < 20) # (True) if anyone is true.
print(a < 5 or b < 20) # (False) if both are true.

```

```python
# Not Operator
print(not b < 20) # Reverse the result.

```
