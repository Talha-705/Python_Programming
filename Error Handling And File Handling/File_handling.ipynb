# File_Handling

## There are 6 access modes in Python:

1- Read Only (‘r’)

2- Read and Write (‘r+’)

3- Write Only (‘w’)

4- Write and Read (‘w+’)

5-Append Only (‘a’)

6-Append and Read (‘a+’)


### 1. READ MODE

```python
file = open('data.txt')
print(file)
```

case no 1

```python
# By default, The file opens in read mode
print(file.readline())
print(file.readline())
print(file.readline())
```

Case No 2

```python
file = open('data.txt')
print(file)
print(file.readlines())
```

Case NO 3

```python
file = open('data.txt')
print(file)
print(file.readlines()[0])
```

Case No 4


```python
file = open('data.txt')
print(file)
print(file.readline())
file.close()
```

```python
print(file.readline())
```

## 2. WRITE MODE


```python
with open("data.txt", 'w') as file:  
    file.write("Pakistan zinda bad")
```

3. APPEND MODE


```python
with open("data.txt", 'a') as file:
    file.write("\nPakistan zinda bad")
```
