## 3. Constructor Function
A constructor is a special method that runs automatically when an object is created. In Python, we use __init__() as the constructor.

#### Important Points
- __init__() initializes the object's data.
- self refers to the current object.
- The constructor runs automatically when we create an object.

```python
class Person():
    def __init__(self, name, age): # constructor method
        self.name = name # object attriburtes
        self.age = age # object attriburtes
    
    def walk(self): # object methods
        print('The person is walking')
    def talk(self):
        print(f'{self.name} is talking')

    def eat(self):
        print(f'{self.name} is eating')
```

```python
p1 = Person('Ali', 21)
```

```python
p1.name
```

```python
p1.age
```

```python
p1.walk()
```

```python
#  now we will be  creating a second object of the class Person and calling its methods and attributes.
p2 = Person('Ahmed', 25)
```

```python
print(f"name of person of object one is:  {p1.name}")
```

```python
print(f"name of person through object two is:  {p2.name}")
```

## 4. Destructor
A destructor is a special method that is called when an object is being destroyed. In Python, we use __del__() as the destructor.

#### Important Points
- __del__() is the destructor method.
- It may run when an object is destroyed or garbage-collected.
- del s1 removes the reference named s1; it does not always destroy the object immediately if other references exist.

```python
class Person:
    # Constructor
    def __init__(self, name):
        self.name = name
        print(f"Constructor: Hello, my name is {self.name}")
    def walk(self): # object methods
        print('The person is walking')

        

    # Destructor
    def __del__(self):
        print(f"Destructor: Goodbye from {self.name}")
```

```python
# Constructor is called when the object is created
p1 = Person('Ali')
```

```python
p1.walk()
```

```python
del p1 # Destructor is called when the object is deleted
```

```python
p1.walk()
```

✅ Main Purpose: Resource Cleanup
When an object is destroyed, a destructor lets you:

🔓 Release external resources (like files, network connections, or database connections)
🧹 Free up memory or do final cleanup before the object is removed from memory
📦 Close open files or streams
🔐 Log messages or actions when an object is deleted


## 4. Inheritance
Inheritance is a feature in which one class acquires the properties and methods of another class.
- Parent class: The class whose properties are inherited.
- Child class: The class that inherits properties from the parent class.

#### Uses of Inheritance in Python OOP
1. Code Reusability: Reuse code from a parent class without writing it again.
2. Reduce Code: Avoid repeating the same code in different classes.
3. Easy Maintenance: Update common code in the parent class.
4. Extend Features: Add new methods and properties in the child class.
5. Organize Code: Create relationships between classes.

#### Types of Inheritance
1. Single inheritance
2. Multiple inheritance
3. Multilevel inheritance
4. Hierarchical inheritance
5. Hybrid inheritance

```python
class Person():
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
        
    def walk(self):
        print('He is Walking...')

class Student(Person):
    roll_no = 1231
    def register_course(self):
        print('Computer Science course is registered')
```

```python
student1 = Student('Talha', 22, 'male')
```

```python
print('The age of Student:', student1.age) # inheritence
print('The name of Student:',student1.name)

print(student1.walk()) # inheritence
print('---------------------------')
print(student1.roll_no)
```

### 1. Multilevel Inheritance
A class inherits from a child class, forming a chain of inheritance.

```python
# Base class
class Animal:
    def speak(self):
        print("Animal speaks")

# Intermediate class (inherits from Animal)
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# Derived class (inherits from Dog)
class Puppy(Dog):
    def weep(self):
        print("Puppy weeps")
```

```python
# Creating object
p = Puppy()
p.speak()  # From Animal
p.bark()   # From Dog
p.weep()   # From Puppy
```

### 2. Hierarchical Inheritance
Multiple child classes inherit from the same parent class.

```python
# Base class
class Animal:
    def speak(self):
        print("Animal speaks")

# Derived class 1
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# Derived class 2
class Cat(Animal):
    def meow(self):
        print("Cat meows")
```

```python
# Creating objects
d = Dog()
d.speak()
d.bark()
```

```python
# Creating objects
c = Cat()
c.speak()
c.meow()
```

### 3. Multiple Inheritance
One child class inherits from two or more parent classes.

```python
# First parent class
class Father:
    def skills(self):
        print("Father: Gardening")

# Second parent class
class Mother:
    def skills(self):
        print("Mother: Cooking")

# Child class inheriting from both
class Child(Father, Mother):
    def own_skill(self):
        print("Child: Painting")
```

```python
# Creating object
c = Child()
c.skills()       # Shows Father’s skill (Python follows left-to-right order)
c.own_skill()    # Child's own method
```

```python
# For getting only mother skills
Mother.skills(c)
```

```python
# if want to get both mother and father skills 
class Father:
    def skills(self):
        print("Father: Gardening")

class Mother:
    def skills(self):
        print("Mother: Cooking")

class Child(Father, Mother):
    def skills(self):
        Father.skills(self)
        Mother.skills(self)

    def own_skill(self):
        print("Child: Painting")

c = Child()
c.skills()
c.own_skill()
```
